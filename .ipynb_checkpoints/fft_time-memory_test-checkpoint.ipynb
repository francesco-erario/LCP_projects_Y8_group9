```python
import numpy as np
import scipy.signal as scs
import matplotlib.pyplot as plt
import time # library to measure time
import tracemalloc # library to measure memory usage
import pandas as pd

np.random.seed(42) # set fixed seed
```

```python
def evolution(world, neighbours):
    new_world = np.zeros_like(world)
    # condition1 = ( neighbours < 2 ) useless, we start from a 0's new world ——> dead
    # condition2 = ( neighbours > 3 ) ——> dead
    # condition3 = ( neighbours == 2 ) | ( neighbours == 3 ) # '|' is bitwise 'or' ——> alive only if already alive
    # condition4 = ( neighbours == 3 ) ——> become alive if dead
    survive = ( world == 1 ) & ( neighbours == 2 ) # does not need the 3 ——> always alive
    born = ( neighbours == 3 )
    
    new_world[survive | born] = 1

    return new_world
```

```python
size_list = [10, 30, 100, 150, 330, 640, 1000, 1500, 3000]

data = []

for idx, N in enumerate(size_list):

    tracemalloc.start()
    start_time = time.time()

    # Initialize the world at random
    world = np.random.randint(0, 2, size=(N, N))

    # Kernel 3x3 (center excluded)
    kernel = np.ones((3, 3))
    kernel[1, 1] = 0

    # --- COSTRUZIONE KERNEL PERIODICO PER FFT ---
    kernel_padded = np.zeros((N, N))
    kernel_padded[:3, :3] = kernel

    # Shift per centrare il kernel (necessario per convoluzione corretta)
    kernel_padded = np.roll(kernel_padded, -1, axis=0)
    kernel_padded = np.roll(kernel_padded, -1, axis=1)

    # FFT del kernel (si può precomputare fuori dal ciclo temporale)
    kernel_fft = np.fft.fft2(kernel_padded)

    # Temporary cycle
    T = 10
    for t in range(T):

        world_fft = np.fft.fft2(world)

        # Convoluzione circolare via FFT
        convolution = np.fft.ifft2(world_fft * kernel_fft).real

        # Arrotondamento (errori numerici floating point)
        convolution = np.rint(convolution).astype(int)

        world = evolution(world, convolution)

    end_time = time.time()
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    data.append({
        'N': N,
        'time_s': end_time - start_time,
        'current_kB': current / 1024,
        'peak_kB': peak / 1024
    })

    print(f"——— Iteration: {idx} –––")
    print(f"Time elapsed: {end_time - start_time:.5f} s")
    print(f"Current memory: {current / 1024:.5f} kB")
    print(f"Peak memory: {peak / 1024:.5f} kB")
    print("\n")

df = pd.DataFrame(data)
df.to_csv('fft_periodic_convolution_TM_data.csv', index=False)
```
