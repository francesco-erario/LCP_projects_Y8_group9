```python
def save_pattern_gif(pattern_func, cell_size=8, fps=12):
    
    pattern = pattern_func()
    ph, pw = pattern.shape
    
    # --- Automatic choice of steps ---
    name = pattern_func.__name__
    
    if name in ["pattern_blinker", "pattern_toad", "pattern_beacon"]:
        steps = 20
        
    elif name in ["pattern_block", "pattern_boat", 
                  "pattern_tub", "pattern_beehive", 
                  "pattern_loaf"]:
        steps = 20
        
    elif name in ["pattern_glider", "pattern_lwss", 
                  "pattern_mwss", "pattern_hwss"]:
        steps = 120
        
    elif name in ["pattern_pulsar", "glider_gun"]:
        steps = 150
        
    else:
        steps = 100
    # -------------------------------------
    
    # Automatic dimension of the grid
    N = max(ph, pw) + 10
    world = np.zeros((N, N), dtype=int)
    
    # Pattern centering
    start_x = (N - ph) // 2
    start_y = (N - pw) // 2
    world[start_x:start_x+ph, start_y:start_y+pw] = pattern
    
    game = GameOfLife(N, world)
    
    fig, ax = plt.subplots(figsize=(cell_size, cell_size))
    ax.set_axis_off()
    im = ax.imshow(game.world, cmap='binary')

    def update(frame):
        game.step()
        im.set_data(game.world)
        return [im]

    anim = animation.FuncAnimation(
        fig,
        update,
        frames=steps,
        blit=True
    )
    
    filename = name + ".gif"
    
    anim.save(
        filename,
        writer=PillowWriter(fps=fps),
        dpi=150
    )
    
    plt.close(fig)
    print(f"Salvata: {filename} ({steps} steps)")
```

```python
patterns = [
    pattern_glider,
    pattern_blinker,
    pattern_block,
    pattern_beehive,
    pattern_loaf,
    pattern_boat,
    pattern_tub,
    pattern_toad,
    pattern_beacon,
    pattern_lwss,
    pattern_mwss,
    pattern_hwss,
    pattern_pulsar,
    glider_gun
]

for p in patterns:
    save_pattern_gif(p)
```
